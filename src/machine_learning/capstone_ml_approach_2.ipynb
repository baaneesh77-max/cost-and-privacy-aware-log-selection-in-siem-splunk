```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

# 1. Load dataset
df = pd.read_csv(r"D:\Documents\capstone-ml\dataset\master_dataset1.csv",encoding="cp1252", on_bad_lines="skip")

# 2. Clean column names
df.columns = df.columns.str.strip()

# 3. Keep only verbose logs
df_verbose = df[df["log_version"] == "verbose"].copy()

# 4. Remove spaces in text values
for col in df_verbose.columns:
    if df_verbose[col].dtype == "object":
        df_verbose[col] = df_verbose[col].astype(str).str.strip()

# 5. Drop columns not needed
drop_cols = ["attack_type", "log_version", "raw_log"]
drop_cols = [col for col in drop_cols if col in df_verbose.columns]

X = df_verbose.drop(columns=drop_cols)
y = df_verbose["attack_type"]

# 6. Convert text columns into numbers
X = pd.get_dummies(X, dummy_na=True)

# 7. Fill missing values
X = X.fillna(0)

# 8. Split into train and test
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

# 9. Train model
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# 10. Predict
y_pred = model.predict(X_test)

# 11. Results
print("VERBOSE LOG RESULTS")
print("Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

# 1. Load dataset
df = pd.read_csv(r"D:\Documents\capstone-ml\dataset\master_dataset1.csv")

# 2. Clean column names
df.columns = df.columns.str.strip()

# 3. Keep only reduced logs
df_reduced = df[df["log_version"] == "reduced"].copy()

# 4. Remove spaces in text values
for col in df_reduced.columns:
    if df_reduced[col].dtype == "object":
        df_reduced[col] = df_reduced[col].astype(str).str.strip()

# 5. Drop columns not needed
drop_cols = ["attack_type", "log_version", "raw_log"]
drop_cols = [col for col in drop_cols if col in df_reduced.columns]

X = df_reduced.drop(columns=drop_cols)
y = df_reduced["attack_type"]

# 6. Convert text columns into numbers
X = pd.get_dummies(X, dummy_na=True)

# 7. Fill missing values
X = X.fillna(0)

# 8. Split into train and test
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

# 9. Train model
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# 10. Predict
y_pred = model.predict(X_test)

# 11. Results
print("REDUCED LOG RESULTS")
print("Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

```python
feature_importance = pd.DataFrame({
    "feature": X.columns,
    "importance": model.feature_importances_
})

feature_importance = feature_importance.sort_values(by="importance", ascending=False)

print(feature_importance.head(20))
```
